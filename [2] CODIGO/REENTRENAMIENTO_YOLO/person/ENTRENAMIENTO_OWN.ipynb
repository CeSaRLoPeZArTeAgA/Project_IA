La estructura del set de datos (dataset_vehiculos) es:


- 399 imágenes de entrenamiento
- 50 imágenes de validación
- 50 imágenes de prueba

```python
from ultralytics import YOLO
import os
import yaml

# =========== DESCARGA MODELO Y MANEJO DE PATH ========================

# Obtener el directorio actual del script
script_dir = os.getcwd()
print("[INFO] Ruta del directorio donde esta el script: ",script_dir)

# Ruta local al modelo YOLO a descargar
model_path = os.path.join(script_dir, "yolo11n.pt")
print("[INFO] Ruta del modelo donde se descargara: ",model_path)

# ruta de la data para uso de yaml
path_data=os.path.join(script_dir,"DATA_DETECCION","dataset_vehiculos")
print("[INFO] Ruta del archivo de la data para reentrenamiento: ",path_data)

# ruta del archivo yaml
path_yaml=os.path.join(path_data, "data.yaml")
print("[INFO] Ruta del archivo yaml: ",path_yaml)

print("\n\n")


# ========== CREACION DE DICCIONARIO P/BUSQUEDA DE LA DATA ===============

data = {
    'path': path_data, 
    'train': 'images/train',
    'val': 'images/val',
    'test': 'images/test',
    'names': { 0: 'car'}
}


with open(path_yaml, 'w',encoding='utf-8') as file:
    yaml.dump(data, file,default_flow_style=False,sort_keys=False)



# ========== DESCARGA DE MODELO Y REENTRENAMIENTO DEL MODELO ===============

# cargar modelo YOLO 11 nano pt(deteccion)
model = YOLO("yolo11n.pt")

# Afinar el modelo
results = model.train(
                    data=path_yaml,
                    epochs=10, # Número de iteraciones
                    imgsz=640, # Tamaño de las imágenes (máxima dimensión)
                    plots=True, # Guardar gráficos de entrenamiento
                    project=path_data,  # tu carpeta personalizada
                    name="modelo_reentreno"  # nombre del subdirectorio (por defecto sería "train")
                    )
                    



```
